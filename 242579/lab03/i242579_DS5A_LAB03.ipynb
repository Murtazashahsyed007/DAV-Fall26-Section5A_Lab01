```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Superstore.csv',encoding='latin1')
df
```

```python
df.head()
```

```python
df.columns
```

# Task1

```python
df['months']=pd.to_datetime(df['Order Date']).dt.month_name()
df['months']
```

```python
monthly_sales=df.groupby('months')['Sales'].sum()
monthly_sales
```

```python

sns.barplot(x=monthly_sales.index,y=monthly_sales.values)
plt.title('Monthly Sales')
plt.xlabel('Months')
plt.xticks(rotation=90)
plt.ylabel('Sales')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.lineplot(x=monthly_sales.index,y=monthly_sales.values)
plt.title('Monthly Sales')
plt.xlabel('Months')
plt.xticks(rotation=90)
plt.ylabel('Sales')
plt.show()
```

```python
plt.figure(figsize=(10,10))
plt.pie(monthly_sales,labels=monthly_sales.index,autopct='%1.2f%%')
plt.title('Monthly Sales')
plt.show()
```

```python
#I would prefer Pie Chart as it shows the distribution of monthly sales in more accurate and precise manner
```

```python
region_sales=df.groupby('Region')['Sales'].sum()
region_sales
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=region_sales.index, y=region_sales.values, palette='viridis')
plt.title('Category Wise Sales')
plt.xlabel('Category')
plt.xticks(rotation=45)
plt.ylabel('Sales')
plt.ylim(20000,)
plt.tight_layout()
plt.show()
```

```python
category_sales = df.groupby('Category')['Sales'].sum()
category_profit = df.groupby('Category')['Profit'].sum()
```

```python
fig, ax1 = plt.subplots(figsize=(12, 7))

sns.lineplot(x=category_sales.index, y=category_sales.values, ax=ax1, color='skyblue', label='Sales')
ax1.set_xlabel('Category')
ax1.set_ylabel('Total Sales', color='skyblue')
ax1.tick_params(axis='y', labelcolor='skyblue')
ax1.tick_params(axis='x', rotation=45)

ax2 = ax1.twinx()
sns.lineplot(x=category_profit.index, y=category_profit.values, ax=ax2, color='red', marker='o', label='Profit')
ax2.set_ylabel('Total Profit', color='red')
ax2.tick_params(axis='y', labelcolor='red')

plt.title('Category Wise Sales and Profit')
fig.legend(loc='upper right', bbox_to_anchor=(1, 1), bbox_transform=ax1.transAxes)

plt.tight_layout()
plt.show()
```

```python
sns.barplot(x=monthly_sales.index,y=monthly_sales.values,palette='coolwarm')
plt.title('Monthly Profit')
plt.xlabel('Months')
plt.xticks(rotation=90)
plt.ylabel('Sales')
plt.gca().invert_yaxis()
plt.show()
```
