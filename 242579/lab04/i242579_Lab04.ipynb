# Task 6

```python
df['months']=pd.to_datetime(df['Order Date']).dt.month_name()
df['months']
```

```python
monthly_sales=df.groupby('months')['Sales'].sum()
monthly_sales
```

```python
monthly_rolling=monthly_sales.rolling(window=3).mean()
monthly_rolling
```

```python
plt.figure(figsize=(10,6))
sns.lineplot(x=monthly_sales.index,y=monthly_sales.values)
sns.lineplot(x=monthly_rolling.index,y=monthly_rolling.values)
plt.title('Monthly Sales')
plt.xlabel('Months')
plt.xticks(rotation=90)
plt.ylabel('Sales')
plt.show()
```

# Task 7


```python
df['profitable']=df['Profit']>0
df['profitable']
```

```python
g = sns.FacetGrid(df, col='Region', hue='profitable', col_wrap=2, height=4, aspect=1.2)
g.map_dataframe(sns.scatterplot, x='Discount', y='Profit', s=100, alpha=0.7)
g.add_legend()
g.set_axis_labels('Discount', 'Profit')
g.set_titles(col_template='{col_name} Region')
plt.suptitle('Discount vs. Profit by Region and Profitability', y=1.02)
plt.tight_layout()
plt.show()
```

From the plots, it appears that the **Central** region shows a distinct pattern where higher discounts often lead to significant losses, more so than in other regions. While all regions experience losses with higher discounts, the magnitude and frequency of negative profit at high discount rates seem more pronounced in the Central region.
